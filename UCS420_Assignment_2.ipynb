**Question 1 – Working with Lists

**Question 2 – Working with Tuples
Q2(i) – Find Highest & Lowest**

```python
scores = tuple(l[:8])

print(scores)

print("Max:", max(scores),
      "Index:", scores.index(max(scores)),
      "Count:", scores.count(max(scores)))

print("Min:", min(scores),
      "Index:", scores.index(min(scores)),
      "Count:", scores.count(min(scores)))
```

Q2(ii) – Reverse the Tuple

```python
#as tuples are immutable so swaping is not allowed
reversed_scores = list(scores[::-1])
print(reversed_scores)
```

Q2(iii) – Search a Score

```python
score = int(input("Enter score "))
if scores.count(score):
  print(scores.index(score))
else:
  print("Not present in the tuples")
```

Q2(iv) – Tuple Immutability

```python
try :
  scores[0]=60
except TypeError as e:
  print("Error : ",e)
  print("""You can't randomly edit the tuple like list as tuple are immutable,
  that's why they are also called frozen list.""")
```

Q2(v) – Tuple Unpacking

```python
first_score, second_score, *remaining_scores = scores

print("First Score:", first_score)
print("Second Score:", second_score)
print("Remaining Scores:", remaining_scores)
```

Question 3 – Random Number Analysis
Q3(i) – Generate Random Numbers

```python
import random
random.seed(1024170310)
random_list = [random.randint(100, 900) for _ in range(100)]
print(random_list)
```

Q3(ii) – Count Odd Numbers

```python
print([i for i in random_list if i%2 != 0])
print(len([i for i in random_list if i % 2 != 0]))
```

Q3(iii) – Count Even Numbers

```python
print([i for i in random_list if i%2 == 0])
print(len([i for i in random_list if i % 2 == 0]))
```

Q3(iv) – Find Prime Numbers

```python
def isPrime(n):
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
print(len([i for i in random_list if isPrime(i)]))
prime_list = list(i for i in random_list if isPrime(i));
print(prime_list)
```

Q3(v) – Most Frequent Number

```python
count_list = [random_list.count(i) for i in random_list]
max_count = max(count_list)
for i in range(len(random_list)):
    if count_list[i] == max_count and random_list[i] not in random_list[:i]:
        print(random_list[i], "occurs", max_count, "times")
```

Question 4 – Working with Sets

Q4(i) – Create the Sets

```python
num = 1024170310
l1 = []
l2 = []
while(num):
  digi = num%10
  num//=10
  l1.append(digi*7)
  l2.append(digi*9)
s1 = set(l1[:8])
s2 = set(l2[:8])
print(s1,s2)
```

Q4(ii) – Union

```python
print(s1.union(s2))
```

Q4(iii) – Intersection

```python
print(s1.intersection(s2))
```

Q4(iv) – Difference

```python
print(s1.difference(s2),s2.difference(s1))
# a.difference(b) means a-b on the other hand a.symmetric_difference() means (a-b)U(b-a)
```

Q4(v) – Symmetric Difference

```python
print(s1.symmetric_difference(s2))
```

Q4(vi) – Subset & Superset

```python
print(s1.issubset(s2))
print(s2.issuperset(s1))
```

Q4(vii) – Safe Element Removal

```python
x = int(input("Enter Number"))
s1.discard(x)
# if .remove(x) is used and x is not in set then it will throw an error so if you are unsure about presence of the x in set use discard
print(s1)
```
