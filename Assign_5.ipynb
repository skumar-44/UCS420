Q.3 For the given 2-D array arr=np.array([10, 20, 30], [40, 50, 60], [70, 80, 90]), access elements using row and column indices as follows: 

a) Access 1st row, 2nd column  

b) Access 3rd row, 1st column 

```python
uniq,idx,counts = np.unique(x,return_counts=True,return_index=True)
most_freq = [uniq[i] for i in range(len(uniq)) if counts[i] == np.max(counts)]
print(f'Most freq element : {[i.item() for i in most_freq]}')
print([i for i in range(len(x)) if x[i] in  most_freq])
```

```python
uniq,idx,counts = np.unique(y,return_counts=True,return_index=True)
most_freq = [uniq[i] for i in range(len(uniq)) if counts[i] == np.max(counts)]
print(f'Most freq element : {[i.item() for i in most_freq]}')
print([i for i in range(len(y)) if y[i] in  most_freq])
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>. 

```python
ucs420_sparsh=np.array([10,20,30,40,50,60,70,80,90,15,20,35]).reshape(3,4).copy()
print(f'Mean : {np.mean(ucs420_sparsh)}')
print(f'Median : {np.max(ucs420_sparsh)}')
print(f'Min : {np.min(ucs420_sparsh)}')
print(f'Unique elemnts : {np.unique(ucs420_sparsh)}')
reshaped_ucs420_sparsh=ucs420_sparsh.reshape(4,3)
print(f'New Array : \n{reshaped_ucs420_sparsh}')
resized_ucs420_sparsh=ucs420_sparsh.copy()
resized_ucs420_sparsh.resize(2,3)
print(f'New Array : \n{resized_ucs420_sparsh}')


```
