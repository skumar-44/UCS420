Q4. A simple cognitive-assistive system wants to predict a user's assistance score based on three factors: • Sleep hours  • Activity level  • Stress level  Consider: y = np.array([40, 65, 30, 85]) X = np.array([ [6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8] ]) Here, each row represents one user and each column represents a feature. Perform the following: 

a. Display the shape and dimensions of X. 

b. Calculate: X.T and explain what the transpose represents. 

c. Calculate the matrix product: X.T @ X 

d. Calculate the matrix inverse using: np.linalg.inv() where appropriate. 

e. Implement the Ordinary Least Squares equation: 𝛽 = (𝑋𝑋) 𝑋𝑦 using NumPy. 

f. Explain what the resulting coefficients represent in the context of the three user features. 

g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to obtain a predicted assistance score. 

```python

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dims of X:", X.ndim)
```

```python
XT = X.T

print(XT)
```

```python
XTX = X.T @ X

print(XTX)
```

```python
XTX_inv = np.linalg.inv(XTX)

print(XTX_inv)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("Coeff:", beta)
```

```python
# The coefficients tell us how each of the three 
# factors affects the predicted assistance score 
# when the other factors are kept constant. The 
# coefficient for sleep hours is -5.83, for activity level 
# is 0.5, and for stress level is 7.5. Therefore, according 
# to this model, a one-unit increase in sleep hours decreases 
# the predicted score by about 5.83, while a one-unit increase 
# in activity level or stress level increases the 
# predicted score by 0.5 and 7.5, respectively.
```

```python
new_user = np.array([5, 40, 7])
```

```python
prediction = new_user @ beta

print("Predicted score:", prediction)
```
